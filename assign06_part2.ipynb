```python
class Car:
  model = "Toyota"
  color = "White"
  year = "1987"

  def __init__(self, model, color, year):
    self.model = model
    self.color = color
    self.year = year

  def start(self):
    print(f"this car was created in year {self.year} and model was {self.model} and color was {self.color}")

car = Car("Toyota", "White", "1987")
car.start()
```

```python
class Person:
  def __init__(self, name, age,gender):
    self.name = name
    self.age = age
    self.gender = gender

  def talk(self):
    print(f"Hi, I am {self.name}")

person = Person("John", 30,"F")
person.talk()
```

```python
class Book:
  def __init__(self,name,author,price):
    self.name = name
    self.author = author
    self.price = price

  def printdetail(self):
    print(f"the book name is {self.name} author is {self.author} and price is f{self.price}")
books = Book("Python", "John", 100)
books.printdetail()
```
